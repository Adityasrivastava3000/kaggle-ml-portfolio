```python
from github import Github
from google.colab import userdata
import base64
import os

# Get GitHub token securely from Colab Secrets
github_token = userdata.get("GITHUB_TOKEN")

# Connect to GitHub
g = Github(github_token)

# Get your GitHub account
user = g.get_user()

# Open your repository
repo = user.get_repo("kaggle-ml-portfolio")

print("Connected to:adityasrivastava3000/kaggle-ml-portfolio", repo.full_name)
```

```python
def upload_file_to_github(local_path, github_path, commit_message):
    """
    Upload or update a file in the GitHub repository.
    """

    # Read local file
    with open(local_path, "rb") as f:
        file_content = f.read()

    # Check whether file already exists
    try:
        existing_file = repo.get_contents(github_path)

        repo.update_file(
            path=github_path,
            message=commit_message,
            content=file_content,
            sha=existing_file.sha
        )

        print(f"Updated: {github_path}")

    except Exception:
        repo.create_file(
            path=github_path,
            message=commit_message,
            content=file_content
        )

        print(f"Created: {github_path}")
```
